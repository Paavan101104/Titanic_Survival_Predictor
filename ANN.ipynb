```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import tensorflow 
import streamlit as st
```

## DATA GATHERING

```python
data=pd.read_csv('huge_1M_titanic.csv')
```

```python
data=data.sample(20000)
```

## DATA PREPROCESING

```python
data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
data.columns
```

```python
data['Embarked'].dropna(inplace=True)
```

```python
data=data[['Survived','Pclass','Sex','Age','SibSp','Parch','Fare','Embarked']]
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
data['Age']=data['Age'].fillna(data['Age'].mean(),inplace=True)
```

```python
data['Embarked']=data['Embarked'].fillna(data['Embarked'].mode()[0],inplace=True)
```

```python
data.head(10)
```

```python
data.info()
```

```python
data['Age']=data['Age'].astype('int32')
data['Survived']=data['Survived'].astype('int32')
data['Pclass']=data['Pclass'].astype('int32')
data['SibSp']=data['SibSp'].astype('int32')
data['Parch']=data['Parch'].astype('int32')
```

```python
data.info()
```

## DATA CLEANING

```python
data['Embarked']=data['Embarked'].replace({'S':'Southampton','C':'Cherbourg','Q':'Queenstown'})
```

```python
data.head(10)
```

```python
data.shape
```

```python
data['Fare']=data['Fare'].astype('int32')
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
label=LabelEncoder()
```

```python
data['Sex']=label.fit_transform(data['Sex'])
```

```python
data.head()
```

```python
ohe=OneHotEncoder(sparse_output=False)
```

```python
Embarked=ohe.fit_transform(data[['Embarked']])
```

```python
Embarked=pd.DataFrame(Embarked,columns=['Embarked_C','Embarked_Q','Embarked_S'],index=data.index)
```

```python
Embarked.head()
```

```python
data=pd.concat([data,Embarked],axis=1)
```

```python
data.drop(columns=['Embarked'],inplace=True)
```

```python
data.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scale=StandardScaler()
```

```python
data.columns
```

```python
num_cols=['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
```

```python
data[num_cols]=scale.fit_transform(data[num_cols])
```

```python
data.head()
```

```python
data.isnull().sum()
```

## TRAIN TEST SPLIT

```python
X=data.drop(columns=['Survived'])
```

```python
X.head()
```

```python
X.isnull().sum()
```

```python
Y=data['Survived']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
x_train,x_valid,y_train,y_valid=train_test_split(x_train,y_train,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

## MODEL BUILDING

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model=Sequential([Dense(128,input_shape=(x_train.shape[1],),activation='relu'),#first hidden layer
            Dense(64,activation='relu'),
           Dense(32,activation='relu'),
           Dense(1,activation='sigmoid')] #output layer
           )
```

```python
model.summary()
```

```python
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
EarlyStopping=EarlyStopping(monitor='val_loss',patience=7,restore_best_weights=True)
```

```python
model.fit(x_train,y_train,validation_data=(x_valid,y_valid),epochs=15,batch_size=3,callbacks=[EarlyStopping])
```

```python
model.save('titanic_model.h5')
```
