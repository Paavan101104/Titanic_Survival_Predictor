```python
from tensorflow.keras.models import load_model
import pandas as pd
import pickle
import tensorflow as tf
```

```python
with open('scaler.pkl','rb') as f:
    scaler=pickle.load(f)

with open('label.pkl','rb') as f:
    label=pickle.load(f)

with open('ohe.pkl','rb') as f:
    ohe=pickle.load(f)
```

```python
model=load_model('titanic_model.h5')
```

```python
user={
    'Pclass':2,
    'Sex':'female',
    'Age':59,
    'SibSp':0,
    'Parch':0,
    'Fare':32,
    'Embarked':'Cherbourg'
}
```

```python
user=pd.DataFrame([user])
```

```python
user
```

```python
new=ohe.transform(user[['Embarked']])
```

```python
new=pd.DataFrame(new,columns=['Embarked_C','Embarked_Q','Embarked_S'])
```

```python
user=pd.concat([user,new],axis=1)
```

```python
user.drop(['Embarked'],axis=1,inplace=True)
```

```python
user
```

```python
user['Sex']=label.transform(user['Sex'])
```

```python
user
```

```python
user.columns
```

```python
cols=['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
```

```python
user[cols] = scaler.transform(user[cols])
```

```python
user
```

```python
y=model.predict(user)
```

```python
y
```

```python
if y[0][0]>0.5:
    print ('There may be a chance of survival')
else:
    print('God May bless you')
```
